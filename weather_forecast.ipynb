# Weather Forecast for Lino Lakes, Minnesota

This notebook retrieves the weather forecast from two reliable sources: the National Weather Service (NWS) and Open-Meteo.

## Import Required Libraries

Import necessary libraries such as requests for API calls and pandas for data handling.

```python
import requests
import pandas as pd
from datetime import datetime
```

## Set Up Location

Define the coordinates for Lino Lakes, Minnesota.

```python
# Coordinates for Lino Lakes, Minnesota
lat = 45.1608
lon = -93.0888
location = "Lino Lakes, Minnesota"
```

## Retrieve Forecast from National Weather Service

Fetch the forecast data from the NWS API.

```python
# Fetch forecast from NWS
headers = {'User-Agent': 'WeatherForecastNotebook/1.0 (example@example.com)'}
point_url = f"https://api.weather.gov/points/{lat},{lon}"
point_response = requests.get(point_url, headers=headers)
if point_response.status_code == 200:
    point_data = point_response.json()
    office = point_data['properties']['cwa']
    gridX = point_data['properties']['gridX']
    gridY = point_data['properties']['gridY']
    forecast_url = f"https://api.weather.gov/gridpoints/{office}/{gridX},{gridY}/forecast"
    nws_response = requests.get(forecast_url, headers=headers)
    if nws_response.status_code == 200:
        nws_data = nws_response.json()
        nws_periods = nws_data['properties']['periods']
    else:
        print(f"Forecast API Error: {nws_response.status_code} - {nws_response.text}")
        nws_periods = []
else:
    print(f"Point API Error: {point_response.status_code} - {point_response.text}")
    nws_periods = []
```

## Retrieve Forecast from Open-Meteo

Fetch the forecast data from the Open-Meteo API.

```python
# Fetch forecast from Open-Meteo
openmeteo_url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&daily=temperature_2m_max,temperature_2m_min,precipitation_probability_max&timezone=America%2FChicago"
openmeteo_response = requests.get(openmeteo_url)
openmeteo_data = openmeteo_response.json()
openmeteo_daily = openmeteo_data['daily']
```

## Display the Forecasts

Parse and display the forecast data from both sources.

```python
# Display NWS Forecast
print("National Weather Service Forecast for", location)
print("=" * 50)
for period in nws_periods[:7]:  # Display first 7 periods
    print(f"{period['name']}: {period['temperature']}°{period['temperatureUnit']}")
    print(f"  {period['detailedForecast']}")
    print()

# Display Open-Meteo Forecast
print("Open-Meteo Forecast for", location)
print("=" * 50)
# Convert Celsius to Fahrenheit
max_temps_f = [round(t * 9/5 + 32, 1) for t in openmeteo_daily['temperature_2m_max']]
min_temps_f = [round(t * 9/5 + 32, 1) for t in openmeteo_daily['temperature_2m_min']]

df = pd.DataFrame({
    'Date': openmeteo_daily['time'],
    'Max Temp (°F)': max_temps_f,
    'Min Temp (°F)': min_temps_f,
    'Precipitation Probability (%)': openmeteo_daily['precipitation_probability_max']
})
print(df)
```
